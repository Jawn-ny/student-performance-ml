```python
fig, ax = plt.subplots()
ax.hist(df["G3"], bins=range(0, 21))

ax.set_xlabel("G3")
ax.set_ylabel("Number of students")
ax.set_title("Distribution of G3")

plt.show()
```

```python
fig, ax = plt.subplots()

ax.scatter(df["absences"], df["G3"])

ax.set_xlabel("Absences")
ax.set_ylabel("G3")
ax.set_title("Absences vs G3")

plt.show()
```

```python
fig, ax = plt.subplots()

ax.scatter(df["failures"], df["G3"])

ax.set_xlabel("Failures")
ax.set_ylabel("G3")
ax.set_title("Failures vs G3")

plt.show()
```

```python
fig, ax = plt.subplots()

ax.scatter(df["studytime"], df["G3"])

ax.set_xlabel("Study Time")
ax.set_ylabel("G3")
ax.set_title("Study Time vs G3")

plt.show()
```

```python
g3_counts = df["G3"].value_counts().sort_index()

fig, ax = plt.subplots()

ax.bar(g3_counts.index, g3_counts.values)

ax.set_xlabel("G3")
ax.set_ylabel("Number of students")
ax.set_title("Number of Students by G3")

plt.show()
```
